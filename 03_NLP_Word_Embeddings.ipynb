# NLP — 08. Word Embeddings

So far, we represented text using BoW and TF-IDF. Both convert words into numbers, but they have a major limitation:

> They treat words mostly as independent tokens and don't really understand relationships between words.

**Word embeddings** are the next step.

## 1. What are Word Embeddings?

A **word embedding** is a way of representing a word as a **dense numerical vector.**

Instead of representing: `king`

as a single ID like: `42`

we represent it with a vector such as:

`[0.21, -0.45, 0.78, 0.12, ...]`

For example, imagine a simplified 3-dimensional embedding:

```
king   → [0.8, 0.7, 0.6]
queen  → [0.8, 0.6, 0.7]
apple  → [0.1, 0.9, 0.2]
car    → [0.2, 0.1, 0.8]
```

Real embeddings usually have `hundreds of dimensions.`

The important idea is not the individual numbers themselves. The important thing is that **similar words tend to have similar vectors.**

## 2. Why do we need embeddings?

Consider these two sentences:

```
I love dogs.
I love puppies.
```

BoW/TF-IDF sees:

```
dogs
puppies
```

as two completely different words.

It doesn't inherently know that:
`dog ≈ puppy`

But an embedding model can learn that words appearing in similar contexts tend to have related meanings.

For example:

```
dog     → [0.72, 0.31, 0.65, ...]
puppy   → [0.70, 0.34, 0.63, ...]
car     → [0.12, 0.81, 0.21, ...]
```

The vectors for **dog** and **puppy** would be relatively close.

That's the beginning of **semantic representation.**

## 3. The key idea: meaning from context

A famous idea in NLP is:

> **You can learn a lot about a word from the words that appear around it.**

Consider:

```
The cat is drinking milk.
The dog is drinking milk.
```

`cat` and `dog` appear in similar contexts.

Another example:

```
I drove my car to work.
I drove my vehicle to work.
```

`car` and `vehicle` appear in similar contexts.

An embedding model learns these patterns from large amounts of text.

So instead of manually telling the model:

```
dog = animal
car = vehicle
```

the model **learns relationships from language data.**

## 4. Sparse vs Dense representations

This is an important distinction.

### BoW / TF-IDF

Suppose our vocabulary contains 10,000 words.

A document might be represented as:

`[0, 0, 0, 1, 0, 0, 0, 0, 2, 0, ...]`

Most values are zero.

This is a **sparse vector.**

### Embeddings

A word might instead be represented as:

`[0.21, -0.43, 0.72, 0.15, -0.09, ...]`

Most dimensions contain meaningful numerical values.

This is a `dense vector.`

So:

```
BoW / TF-IDF
       ↓
Sparse vectors
       ↓
Mostly word occurrence information
```
----
```
Word Embeddings
       ↓
Dense vectors
       ↓
Semantic relationships
```

## 5. What does each dimension mean?

This is where beginners often get confused.

Suppose:

`king → [0.72, 0.45, -0.13, 0.81, ...]`

You might ask:

> Does dimension 1 mean "royalty"?

Usually, **not directly.**

The dimensions of learned embeddings are generally **not individually interpretable.**

The meaning is distributed across many dimensions.

Think of it like a coordinate system.

Instead of:

```
Dimension 1 = royalty
Dimension 2 = gender
Dimension 3 = age
```

it's more like:

```
The entire vector
        ↓
represents the learned characteristics of the word
```

## 6. Similar words → similar vectors

One of the most useful properties of embeddings is that we can compare vectors.

Imagine:

```
king   → vector A
queen  → vector B
apple  → vector C
```

If:

```
distance(A, B) = small
distance(A, C) = large
```

then the model considers `king` and `queen` more related than `king` and `apple`.

A common measure is **cosine similarity.**

The idea is:

`similarity ≈ how close the vectors point in the same direction`

For example:

```
dog ↔ puppy       high similarity
dog ↔ cat         high similarity
dog ↔ airplane    low similarity
```

This becomes extremely important later when we study **embeddings for semantic search and RAG.**

## 7. Word embeddings can capture relationships

Good embedding models can learn relationships such as:

```
king ↔ queen
man ↔ woman
Paris ↔ France
Tokyo ↔ Japan
```

Some early embedding models even exhibited interesting vector relationships such as:

`king - man + woman ≈ queen`

Conceptually:

```
king
  ↓
remove "male" relationship
  ↓
add "female" relationship
  ↓
queen
```

This was one of the reasons Word2Vec became so influential.

But don't think of this as a universal mathematical rule. It is an observed property of some learned embedding spaces, not a guarantee for every embedding model.

## 8. Simple Python example

````python
#Let's create a very small example ourselves.

import numpy as np

embeddings = {
    "dog": np.array([0.8, 0.7, 0.6]),
    "puppy": np.array([0.79, 0.72, 0.61]),
    "cat": np.array([0.75, 0.68, 0.64]),
    "car": np.array([0.1, 0.2, 0.9])
}
````

````python
#We can calculate cosine similarity:

from sklearn.metrics.pairwise import cosine_similarity

dog = embeddings["dog"].reshape(1, -1)
puppy = embeddings["puppy"].reshape(1, -1)
car = embeddings["car"].reshape(1, -1)

print("Dog vs Puppy:",
      cosine_similarity(dog, puppy)[0][0])

print("Dog vs Car:",
      cosine_similarity(dog, car)[0][0])
````

The numbers aren't meaningful because we invented these embeddings. They're just demonstrating the principle.

Real embedding models learn the vectors from data.

## 9. How are embeddings actually learned?

This is the important transition toward **Word2Vec.**

We don't normally sit down and manually create:

`dog → [0.8, 0.7, 0.6]`

Instead, we give a model lots of text.

For example:

```
The dog drinks water.
The dog likes food.
The cat drinks milk.
The cat likes food.
```

The model repeatedly looks at words and their contexts.

Over time, it adjusts the numerical vectors so that words appearing in similar contexts develop similar representations.

Very roughly:

```
Large text corpus
       ↓
Context information
       ↓
Training
       ↓
Learned vectors
       ↓
Word embeddings
```

**Word2Vec** is one of the classic algorithms designed specifically to learn these word embeddings.

## 10. Important distinction: Word Embeddings ≠ Token IDs

This is also connected to our tokenization lesson.

Suppose:
`"dog"`
gets token ID:
`512`

That `512` is not the meaning of dog.

It's simply an identifier.

Then an embedding layer can map:

```
token ID 512
       ↓
[0.72, -0.14, 0.63, ...]
```

So:

```
Token
  ↓
Token ID
  ↓
Embedding vector
```

This distinction becomes **very important when we study Transformers and LLMs.**

# NLP — 09. Word2Vec

In the previous topic, we learned **what word embeddings are.**

Now the natural question is:

> **How do we actually create those word vectors?**

One of the most important early methods for doing this is **Word2Vec.**

## 1. What is Word2Vec?

**Word2Vec** is a neural-network-based technique for learning **word embeddings from text.**

Instead of manually assigning vectors to words, Word2Vec learns them by looking at **which words occur near each other.**

For example:

```
The cat drinks milk.
The dog drinks water.
The cat likes food.
The dog likes food.
```

The model notices that:

```
cat → drinks, milk, likes, food
dog → drinks, water, likes, food
```

Because `cat` and `dog` appear in similar contexts, their learned vectors tend to become similar.

So the basic idea is:

> **Words that occur in similar contexts tend to have similar meanings.**

This is the foundation of Word2Vec.

## 2. The intuition behind Word2Vec

Imagine we have this sentence:

`The cat drinks milk`

Take the word:
`cat`

The surrounding words might be:

`The     drinks`

or, with a larger context window:

`The    drinks    milk`

Word2Vec uses these context words to learn a representation for cat.

Now consider:

`The dog drinks milk`

`dog` occurs in almost the same context.

Therefore:

`cat ≈ dog`

in the learned embedding space.

## 3. Word2Vec doesn't directly learn "meaning"

This distinction is important.

Word2Vec isn't given definitions such as:

```
cat = small animal
dog = animal
```

Instead, it receives text:

```
cat drinks milk
dog drinks milk
```

and learns statistical relationships.

So:

```
Text
 ↓
Context relationships
 ↓
Training
 ↓
Word vectors
```

The meaning-like properties emerge from patterns in language.

## 4. Word2Vec has two architectures

Word2Vec actually refers to two closely related approaches:

**1. CBOW**

**Continuous Bag of Words**

Predict the **target word from surrounding context words.**

`Context → Target`

Example:

The ___ drinks milk
      ↓
     cat

The model tries to predict:
`cat`
from surrounding words.

**2. Skip-gram**

The opposite direction:

`Target → Context`

Given:
`cat`

the model tries to predict nearby words:

```
The
drinks
milk
```

So:

```
CBOW:
Context → Word

Skip-gram:
Word → Context
```

We'll study the difference between them in the **next topic**, because it is important enough to understand separately.

## 5. How Skip-gram works

Let's take:
`The cat drinks milk`

Suppose our target word is:
`cat`

and our context window is `1`.

That means we look one word on either side:

```
The  [cat]  drinks  milk
 ↑             ↑
context      context
```

Our training pairs become:

`(cat, The)`

`(cat, drinks)`

The model learns:

> Given `cat`, predict words that occur nearby.

With a larger window, we could have:

```
(cat, The)
(cat, drinks)
(cat, milk)
```

## 6. How CBOW works

Now reverse the idea.

Using:

`The cat drinks milk`

Suppose `cat` is the target.

The context might be:

`The + drinks`

The model receives:

```
The
drinks
 ↓
?
```

and tries to predict:
`cat`

So:

```
**CBOW**

The       drinks
  \       /
   \     /
    Model
      ↓
     cat
```

Whereas Skip-gram is:

```
**Skip-gram**

     cat
    /   \
   ↓     ↓
 The   drinks
```

## 11. Training Word2Vec in Python

````python
#!pip install gensim
````

````python
from gensim.models import Word2Vec

#Let's create a tiny corpus:

sentences = [
    ["the", "cat", "drinks", "milk"],
    ["the", "dog", "drinks", "water"],
    ["the", "cat", "likes", "food"],
    ["the", "dog", "likes", "food"],
    ["the", "cat", "is", "small"],
    ["the", "dog", "is", "friendly"]
]

#Train Word2Vec

model = Word2Vec(
    sentences,      #our training corpus
    vector_size=50, #each word gets a 50-dimensionalvector
    window=2,       #look at nearby words within a window of 2
    min_count=1,    #include words appearing at least once
    workers=1       #number of CPU workers
)
````

## 12. Getting a word's embedding

````python
vector = model.wv["cat"]
print(vector)
````

## 13. Finding similar words

This is one of the most useful features.

````python
similar_words = model.wv.most_similar("cat")
print(similar_words)
````

## 14. Word2Vec limitations

### 1. One word → one vector

The word:
`bank`
gets one embedding.

But:

`I deposited money in the bank.`

and:

`I sat beside the river bank.`

have different meanings.

Word2Vec doesn't dynamically change the embedding based on the sentence.

### 2. It doesn't understand full sentences

Word2Vec focuses primarily on relationships between individual words and their contexts.
It isn't a modern language model capable of generating or understanding arbitrary text the way an LLM does.

### 3. Limited handling of rare/unseen words

Traditional Word2Vec uses a fixed vocabulary learned during training.
A word that wasn't included in the vocabulary may not have a usable embedding.

## 15. The bigger NLP story

That's why **Word2Vec is worth learning even though modern LLMs don't simply work like Word2Vec.**

## Remember

The most important points are:

**1. Word2Vec learns word embeddings from word-context relationships.**

**2.** It has two approaches:
   - **CBOW:** context → target word
   - **Skip-gram:** target word → context words

**3.** Words appearing in similar contexts tend to get similar vectors.

**4.** The learned vectors can capture semantic relationships.

**5.** Word2Vec gives a **static embedding:** one word generally has one vector.

**6.** That last limitation leads us toward **contextual embeddings.**

## 1. Start with one sentence

## 2. CBOW — Continuous Bag of Words

CBOW works like this:

> **Use surrounding words to predict the missing/target word.**

Suppose the target is:

## 3. Skip-gram

Skip-gram does the opposite.
> ***Use the target word to predict surrounding context words.**

Take:

`The [cat] drinks milk`

## 4. The easiest way to remember

Think of a sentence:

`The cat drinks milk`

**CBOW asks:**

> **"What word is missing here?"**

**Skip-gram asks:**

> **"What words usually appear around this word?"**

## 5. Why is it called "Continuous Bag of Words"?

The **bag of words** part means the model uses the surrounding words as a group rather than relying heavily on their order.

For example, the context:

`The + drinks`

is treated as the context for predicting:

`cat`

The word **continuous** refers to the continuous/dense vector representation used by the neural network, rather than traditional discrete word representations.

Don't confuse CBOW with the traditional **Bag of Words (BoW)** technique we studied earlier.

They are related in terminology but are **not the same algorithm.**

## 6. Which one is faster?

Generally:

**CBOW**

CBOW predicts **one target from several context words.**

So it tends to be:
- computationally efficient
- faster to train
- effective for frequent words

**Skip-gram**

Skip-gram generates multiple target-context pairs.

So training can be more computationally expensive.

But it tends to work well for:
- smaller datasets
- rare words
- learning useful representations for less frequent words

These are **general tendencies,** not absolute rules.

## 7. CBOW vs Skip-gram

| Feature | CBOW | Skip-gram |
|---|---|---|
| Full form | Continuous Bag of Words | Skip-gram |
| Input | Context words | Target word |
| Output | Target word | Context words |
| Direction | Context → Target | Target → Context |
| Training examples | Fewer | More |
| Training speed | Generally faster | Generally slower |
| Frequent words | Works well | Works well |
| Rare words | Can be weaker | Often useful |
| Basic idea | Fill in the missing word | Find words around a word |

## 10. Negative Sampling

There's one more important idea associated with Word2Vec:

**Negative sampling.**

Imagine Skip-gram sees:

`dog → animal`

That's a genuine context relationship.

This makes training much more efficient than calculating probabilities over an enormous vocabulary for every training example.

You don't need to memorize the mathematical details yet. Just understand:

> **Negative sampling gives Word2Vec an efficient way to learn which word pairs belong together and which don't.**
