# NLP — 01. What is Natural Language Processing?

## 1. What is NLP?

Natural Language Processing (NLP) is a field of Artificial Intelligence that enables computers to process, understand, and generate human language.

Humans naturally understand text, but computers work with numbers. NLP provides techniques to convert language into representations that computers can process.

**Simple example**
A human sees:

> **I love this movie.**

and immediately understands that the person has a **positive opinion.**

A computer initially sees only a sequence of characters: 

`I l o v e ...`

NLP techniques transform this text into representations that algorithms can process.
```
"I love this movie"
        ↓
       NLP
        ↓
Numerical representation
        ↓
Machine Learning / Deep Learning
        ↓
Understand / Classify / Generate
```

## 2. What can NLP do?

Common NLP applications include:

- **Sentiment Analysis** → Positive / Negative
- **Text Classification** → Spam / Not Spam
- **Translation** → English → French
- **Text Summarization** → Long text → Short text
- **Named Entity Recognition** → Identify people, places, organizations
- **Question Answering** → Answer questions from text
- **Chatbots** → Human ↔ Computer
- **Text Generation** → Generate new text

## 3. Why is human language difficult for computers?

This is an important question because it explains why NLP evolved through many different approaches.

### Ambiguity

Consider:

>"I went to the bank."

What does **bank** mean?

- Financial institution?
- River bank?

The meaning depends on context.

### Same meaning, different words

These sentences essentially express the same idea:
```
I am happy.
I am feeling good.
I am delighted.
I am very pleased.
```

A useful NLP system needs to recognize their semantic similarity.

### Word order matters

Compare:

`Dog bites man.`

`Man bites dog.`

The same words are present, but the meaning is completely different.

### Context matters

Consider:

`The car is parked outside because it is broken.`

What does "**it**" refer to?

The car.

Understanding such relationships becomes increasingly important as we move toward Transformers.

We'll follow this progression throughout this NLP section.




## 5. Traditional NLP vs Modern NLP

Traditional NLP often used techniques such as:

- Text preprocessing
- Tokenization
- Bag of Words
- TF-IDF
- Word2Vec

Modern NLP increasingly uses:

- Tokenization
- Embeddings
- Transformers
- Large Language Models

The traditional techniques are still important because they help us understand **how NLP evolved into modern GenAI.**

## 8. Important distinction for GenAI

You should **not assume that all modern LLM pipelines use traditional preprocessing techniques.**

For example, techniques such as:

- stop-word removal
- stemming
- lemmatization
- Bag of Words
- TF-IDF

are extremely useful for understanding **traditional NLP**, but modern LLMs generally use **tokenization + embeddings + Transformer architectures** instead.

That's why we're learning them historically.

The progression matters:

```
Bag of Words
     ↓
TF-IDF
     ↓
Word Embeddings
     ↓
Word2Vec
     ↓
Contextual Embeddings
     ↓
Attention
     ↓
Transformers
     ↓
    LLMs
```

## ✅Key Takeaways

Points to remember:

> **NLP = using computational methods to process human language.**

- Human language is ambiguous and context-dependent.
- Computers need numerical representations of text.
- NLP evolved from rule-based methods → statistical methods → ML → embeddings → neural networks → Transformers → LLMs.
- Traditional NLP techniques such as BoW and TF-IDF are important foundations.
- Modern GenAI primarily relies on **tokenization, embeddings, Transformers, and large-scale pretraining.**
