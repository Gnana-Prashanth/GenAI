# NLP — 01. What is Natural Language Processing?

## 1. What is NLP?

Natural Language Processing (NLP) is a field of Artificial Intelligence that enables computers to process, understand, and generate human language.

Humans naturally understand text, but computers work with numbers. NLP provides techniques to convert language into representations that computers can process.

**Simple example**
A human sees:

> **I love this movie.**

and immediately understands that the person has a **positive opinion.**

A computer initially sees only a sequence of characters: 

`I l o v e ...`

NLP techniques transform this text into representations that algorithms can process.
```
"I love this movie"
        ↓
       NLP
        ↓
Numerical representation
        ↓
Machine Learning / Deep Learning
        ↓
Understand / Classify / Generate
```

## 2. What can NLP do?

Common NLP applications include:

- **Sentiment Analysis** → Positive / Negative
- **Text Classification** → Spam / Not Spam
- **Translation** → English → French
- **Text Summarization** → Long text → Short text
- **Named Entity Recognition** → Identify people, places, organizations
- **Question Answering** → Answer questions from text
- **Chatbots** → Human ↔ Computer
- **Text Generation** → Generate new text

## 3. Why is human language difficult for computers?

This is an important question because it explains why NLP evolved through many different approaches.

### Ambiguity

Consider:

>"I went to the bank."

What does **bank** mean?

- Financial institution?
- River bank?

The meaning depends on context.

### Same meaning, different words

These sentences essentially express the same idea:
```
I am happy.
I am feeling good.
I am delighted.
I am very pleased.
```

A useful NLP system needs to recognize their semantic similarity.

### Word order matters

Compare:

`Dog bites man.`

`Man bites dog.`

The same words are present, but the meaning is completely different.

### Context matters

Consider:

`The car is parked outside because it is broken.`

What does "**it**" refer to?

The car.

Understanding such relationships becomes increasingly important as we move toward Transformers.

# NLP — 02. Text Preprocessing


Before a machine learning or NLP model can work with text, we often need to **clean and prepare the text**.

Think about it this way:

> **Text preprocessing = converting messy human-written text into a cleaner, more consistent form that an NLP system can work with.**

For example, a person might write:

```text
"HEY!!! I REALLY loved this movie 😍😍... Check it out at www.movierulz.com"
```

Depending on our task, we may want to turn this into something like:

```text
"hey i really loved this movie"
```

But there is an important point:

**There is no single preprocessing pipeline that is correct for every NLP problem.**

For traditional NLP, we often performed aggressive cleaning. For modern LLMs, we generally preserve much more of the original text because punctuation, capitalization, formatting, and context can carry meaning.

## 1. Why do we preprocess text?

Raw text can contain many things that make analysis difficult:

```text
Hello!!! 😊
HELLO
hello...
```

To a human, these are obviously related.

To a simple algorithm, they may initially appear different.

Preprocessing can help make text more consistent.

A traditional preprocessing pipeline might look like:

```text
Raw Text
   ↓
Lowercasing
   ↓
Remove unnecessary characters
   ↓
Remove extra spaces
   ↓
Tokenization
   ↓
Stop-word removal
   ↓
Stemming / Lemmatization
   ↓
Clean text
```

We'll study each of these.

## 2. Lowercasing

Suppose we have:

```text
"I Love Python"
"I love python"
"I LOVE PYTHON"
```

A traditional NLP system might treat these as different strings.

We can convert everything to lowercase:

## 3. Removing punctuation

````python
text = "I Love Python"

text = text.lower()

print(text)
````


Now:
```text
Python
PYTHON
python
```

all become: `python`

### Why do this?

It reduces unnecessary variation.

For example, when counting word frequencies:

```text
Python → 1
python → 1
PYTHON → 1
```

could become:

```text
python → 3
```

### But is lowercasing always good?

No.

Consider:

`Apple`

`apple`

**Apple** could refer to the company, while **apple** could refer to the fruit.

Similarly:

```text
US
us
```

have different meanings.

So **lowercasing is task-dependent**.

Consider:

```text
"I love Python!!!"
```

If we're doing simple word-frequency analysis, `!` may not be useful.

We could remove punctuation:

````python
import string

text = "I love Python!!!"

text = text.translate(str.maketrans("", "", string.punctuation))

print(text)
````


Now:

```text
Hello!
Hello.
Hello?
```

could all become: `Hello`

### But again, be careful.

Punctuation can carry meaning.

For example:

`It's`

contains an apostrophe that changes the word structure.

And in sentiment or modern language models:
`What?!`
can convey a different tone from:
`What`

So we shouldn't blindly remove punctuation.

## 4. Removing special characters

Text can contain characters such as: `@ # $ % & * `

Whether these should be removed depends on the application.

For example, in ordinary text classification, we might remove unnecessary symbols.

But imagine analyzing social media:

`#AI`
`@OpenAI`

Here, `#` and `@` are meaningful.

Similarly:

`C++`
`C#`

Removing symbols would completely change the meaning of programming-related text.

So preprocessing is always **task-specific**.

## 5. Removing URLs

Suppose we have:

`"Check this article: https://example.com/article"`

For some NLP tasks, the URL itself isn't useful.

We can replace it with a special token:
`"Check this article: <URL>"`
rather than simply deleting it.

This is often better because we preserve the information that **a URL existed**.

````python
#Example
import re

text = "Visit https://example.com for more information."

text = re.sub(r"https?://\S+|www\.\S+", "<URL>", text)

print(text)
````

## 6. Removing extra whitespace

Raw text may contain unnecessary spaces:
`"I    love     Python"`

We can normalize it:

````python
text = "I    love     Python"

text = " ".join(text.split())

print(text)
````

## 7. Handling numbers

Suppose we have:

`"I bought 3 apples for $100."`

Should we remove the numbers?

It depends.

For some tasks:

```text
3
100
```

may not matter.

But for others, numbers are extremely important.

For example:
`"COVID cases increased from 100 to 500."`

Removing the numbers destroys important information.

So instead of automatically deleting numbers, ask:

> **Does this information matter for my NLP task?**

## 8. Handling emojis

Consider:

```text
"I love this movie 😍"
"I hate this movie 😡"
```

The emojis contain sentiment information.

If we remove them:

```text
"I love this movie"
"I hate this movie"
```

we still have some sentiment information, but we've lost additional context.

For modern NLP systems, emojis can be meaningful tokens.

So again:

> Don't remove something simply because it looks unnecessary.

## 9. Where does Tokenization fit?

This is one of the most important concepts in NLP.

Suppose we have:

`I love Python`

We can split it into smaller pieces:

```text
["I", "love", "Python"]
```

These pieces are called **tokens**.

This process is called **tokenization**.

We'll cover tokenization separately next, because it deserves proper attention.

## 10. Traditional NLP preprocessing vs LLM preprocessing

This distinction is **very important for GenAI**.

In older NLP pipelines, you might see:

```text
Raw text
 ↓
Lowercase
 ↓
Remove punctuation
 ↓
Remove stop words
 ↓
Stem / Lemmatize
 ↓
Bag of Words / TF-IDF
 ↓
ML model
```

With modern LLMs, the process is closer to:

```text
Raw text
 ↓
Tokenizer
 ↓
Token IDs
 ↓
Embeddings
 ↓
Transformer
```

We generally **don't aggressively clean the text before sending it to an LLM**.

Why?

Because the model can learn from things such as:

* punctuation
* capitalization
* word forms
* formatting
* surrounding context
* special tokens

For example, these aren't necessarily equivalent:

```text
STOP
stop
Stop!
```

A modern language model may use those differences as part of understanding the input.

## 11. The most important principle

When you learn preprocessing, don't memorize:

> "Always lowercase."

or:

> "Always remove punctuation."

Instead remember:

> **Preprocessing is task-dependent.**

For a spam classifier, you might clean aggressively.

For sentiment analysis, you might preserve emojis and punctuation.

For an LLM, you generally want to preserve the original text and let the model's tokenizer handle the input.

# NLP — 03. Tokenization

Before a model can process text, we need to break the text into smaller units called **tokens.**

The important thing to understand is that a token is **not necessarily a word.**

## 1. What is Tokenization?

**Tokenization is the process of splitting text into smaller units called tokens.**

For example: `I love Python` can be tokenized into: `["I", "love", "Python"]`

Here, each word is a token.

The basic idea is:
```
Text
 ↓
Tokenization
 ↓
Tokens
```

But there are several ways to tokenize text.

## 2. Word Tokenization

````python
#The simplest approach is to split text into words.

text = "I love Natural Language Processing!"
tokens = text.split()
print(tokens)
````

The problem using `split()` - contains both the word and punctuation.

A more sophisticated tokenizer can treat punctuation separately: `["I", "love", "Natural", "Language", "Processing", "!"]`

## 3. Sentence Tokenization

````python
#Sometimes we don't want individual words. We want to split a document into sentences.

text = "I love Python. NLP is interesting. Transformers are powerful."
sentences = text.split(".")
print(sentences)
````

This is called **sentence tokenization.** It becomes

[

    "I love Python.",
    "NLP is interesting.",
    "Transformers are powerful."
    
]

## 4. Character Tokenization

Instead of splitting into words, we can split text into individual characters.

`hello` becomes: `["h", "e", "l", "l", "o"]`

````python
text = "understanding"
tokens = list(text)
print(tokens)
````

**Advantage:** 
The vocabulary can be very small because there are only a limited number of characters.

**Problem -**
Sequences become very long.

For example: `understanding`
requires many tokens.

The model then has to process much longer sequences.

## 5. Word vs Character Tokenization

Consider: `unbelievable`

**Word tokenization** `["unbelievable"]` - Only one token.

But suppose the model encounters: `unbelievableness`

It may not have that entire word in its vocabulary.

**Character tokenization** `["u", "n", "b", "e", "l", ...]`

The model can represent almost any word, but the sequence becomes much longer.

This leads to an important middle ground:

````python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

#Sample text
txt = "NLTK provides powerful tools for tokenization. It includes word tokenization and sentence tokenization"
print("Sample text:",txt)

#Word tokenization
wt = word_tokenize(txt)
print("\nWord tokenization:", wt)
print("Number of word tokens:",len(wt))

#Sentence tokenization
st = sent_tokenize(txt)
print("\nSentence tokenization:", st)
print("Number of Sentences:",len(st))
````

## 6. Subword Tokenization 

Modern NLP models commonly use **subword tokenization.**

Instead of treating the entire word or every character as a token, we break words into meaningful pieces.

For example, conceptually: `unhappiness`
might become:

`["un", "happiness"]`
or
`["un", "happi", "ness"]`

The exact tokens depend on the tokenizer.

Subword tokenization provides a useful compromise:

```
Subword tokenization
        ↓
Reasonable vocabulary
        +
Reasonable sequence length
```

## 7. Why can't we just have every word in the vocabulary?

Imagine building a vocabulary containing every possible English word.

You immediately run into problems.

**New words**
`ChatGPT`
`OpenAI`
`GenAI`

**Different word forms**
`play`
`plays`
`played`
`playing`

**Misspellings**
`recieve`
`recieved`

**New or specialized words**
`Scientific`, `medical`, `technical`, and `internet terminology` can constantly introduce new words.

A tokenizer using subwords can break unfamiliar words into pieces it already knows.

## 8. Token IDs

There is another important step.

A neural network doesn't understand:
`["I", "love", "Python"]`
directly.

We need to convert tokens into numbers.

Suppose our vocabulary is:

```
"I"       → 101
"love"    → 205
"Python"  → 874
```

Then: `I love Python`
becomes:
`[101, 205, 874]`

These are called token IDs.

So the process becomes:

```
Text
 ↓
Tokenizer
 ↓
Tokens
 ↓
Token IDs
 ↓
Neural Network
```

This is a major concept to remember.

## 9. Special Tokens

Modern tokenizers often use special tokens to provide additional information to the model.

For example, depending on the model:

`<BOS>` can represent `Beginning Of Sequence.`

`<EOS>` can represent `End Of Sequence.`

Other special tokens can indicate things such as:

`<UNK>    → Unknown token`

`<PAD>    → Padding`

`<SEP>    → Separation between sequences`

The exact special tokens depend on the model/tokenizer.

## 10. Tokenization in LLMs

This is where tokenization becomes directly relevant to GenAI.

When you send: `Explain machine learning` to an LLM, the model doesn't directly receive the sentence as English text.

Conceptually:

```
"Explain machine learning"
            ↓
        Tokenizer
            ↓
["Explain", " machine", " learning"]
            ↓
       Token IDs
            ↓
      Neural Network
```

The actual tokens may be different from the illustrative example above.

This is why LLM interfaces often talk about:

> **token limits**

rather than simply word limits.

## 11. Tokens are not equal to words

This is one of the most important things to remember.

Suppose you have: `playing`

It could potentially be represented as: `["play", "ing"]`

So: **1 word ≠ necessarily 1 token**

A sentence containing 100 words might contain more or fewer than 100 tokens depending on the language, vocabulary, punctuation, and tokenizer.

That's why LLM pricing and context windows are generally measured in **tokens.**

## 12. Let's see a real tokenizer

````python
#Hugging Face's transformers library.
from transformers import AutoTokenizer
import warnings

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
text = "I love Natural Language Processing!"
tokens = tokenizer.tokenize(text)
print(tokens)
````

````python
#Now let's convert them into IDs:

token_ids = tokenizer.convert_tokens_to_ids(tokens)
print(token_ids)
#The exact IDs come from the model's vocabulary.
````

## 13. Tokenizer can do everything together

````python
#Instead of manually performing each step:
encoded = tokenizer("i lOVE nlp!")
print(encoded)
````

````python
print(encoded["input_ids"]) # input_ids - These are the numerical representations of the tokens.
print(encoded["attention_mask"]) # attention_mask - This tells the model which positions contain actual input versus padding when sequences are padded.
````

## 14. Different models use different tokenizers

This is important.
There isn't one universal tokenizer.

Different models can tokenize the same sentence differently.

For example:

```
GPT-style model
BERT
LLaMA
Mistral
Gemma
```
    
can have different vocabularies and tokenization algorithms.

So don't memorize specific token IDs.

Understand the process:

```
Text
 ↓
Model-specific tokenizer
 ↓
Tokens
 ↓
Token IDs
 ↓
Model
```

## 15. Tokenization vs Embeddings

Don't confuse these two.

**Tokenization**

Converts text into tokens and then token IDs.

```
"hello"
   ↓
["hello"]
   ↓
[15324]
```

**Embedding**

Converts those token IDs into **dense numerical vectors.**

Conceptually:

```
[15324]
   ↓
[0.21, -0.73, 0.45, 0.18, ...]
```

So:

```
Text
 ↓
Tokenization
 ↓
Token IDs
 ↓
Embeddings
 ↓
Transformer
```

We'll study embeddings later in much more detail.

Earlier we discussed preprocessing:

```

```

# NLP — 04. Stop Words

## 1. What are Stop Words?


Stop words are commonly occurring words that often carry relatively little useful information for certain NLP tasks.

Examples in English include: 
`the,
is,
a,
an,
and,
of,
in,
to,
for,
on,
with`

Consider:

> "The cat is sitting on the table."

A traditional NLP system might consider: `The   is   on   the` less informative than: `cat   sitting   table`

So we may remove the common words.

[stopwords list - nltk github link](https://gist.github.com/sebleier/554280)

````python
#A common NLP library called NLTK - Natural Language Toolkit
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

#Input Sentence
sentence = "This is a sample sentence to show the stop words filteration."

#Case Folding
senetence = sentence.lower()

#Tokenize the sentence
words = word_tokenize(sentence)
print("Tokenized words:", words)

#Filtering stopwords
filtered_sw = [word for word in words if word not in stopwords.words('english')]
print("Sentence after removing stop words:", filtered_sw)

#Check stopwords
stopwords_list = [word for word in words if word in stopwords.words('english')]
print("Stopwords in the sentence:", stopwords_list)
````

## 2. Why remove Stop Words?

Imagine we're building a simple system that needs to determine which words are most important in a large collection of documents.

Suppose we have thousands of documents containing:
`
the,
is,
a,
and,
of
`

These words occur extremely frequently.

If we keep them, they can dominate simple word-frequency-based representations without adding much information.

For example:

`The cat is on the table`
could become:
`["cat", "sitting", "table"]`

This can reduce the number of words the algorithm needs to process.

````python
import nltk
from nltk.corpus import stopwords
nltk.download("stopwords")

stop_words = set(stopwords.words("english"))
print(list(stop_words)[:20])

text = "The cat is sitting on the table"
tokens = text.lower().split()

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]
print("\n",filtered_tokens)
````

## 3. Why shouldn't we always remove stop words?

This is **very important.**

Consider:

> **"I do not like this movie."**

If we remove words such as `"do"` and `"not"`:

`"I like this movie"`

we have completely changed the meaning.

The original sentence is **negative**.

The modified sentence is **positive**.

Another example:

>"The product is not good."

Removing "not" gives:

>"The product is good."

That's obviously a serious problem for sentiment analysis.

## 4. Stop words can carry meaning

Words like:
`not`
`never`
`no`
`without`
can be extremely important.

Consider: 

>**"I like it."**

>**"I do not like it."**

The difference is largely created by the word **"not".**

Therefore, blindly removing stop words can destroy useful information.

## 5. When can stop-word removal be useful?

It can be useful for some **traditional NLP tasks**, particularly when we're primarily interested in content words.

For example, suppose we're building a very simple document search system.

Document:

> "Python is a programming language used for data analysis."

After stop-word removal, we might have:

`["python", "programming", "language", "used", "data", "analysis"]`

This can make simple keyword-based processing more efficient.

It was particularly common with approaches such as:

- Bag of Words
- TF-IDF
- Basic text classification
- Simple information retrieval

## 6. What about modern NLP and LLMs?

This is where you need to make an important distinction.

For traditional NLP:

```
Text
 ↓
Tokenize
 ↓
Remove stop words
 ↓
BoW / TF-IDF
 ↓
ML model
```

For an LLM:

```
Text
 ↓
Tokenizer
 ↓
Token IDs
 ↓
Embeddings
 ↓
Transformer
```

We generally **do not remove stop words before giving text to an LLM.**

Why?

Because modern language models learn from **context and relationships between words.**

Consider:

`"I am happy."`
`"I am not happy."`

The word `"not"` is critical.

Removing it would change the input itself.

So don't carry the traditional rule:

> "Stop words are useless."

The better understanding is:

> **Stop words are common words that may be less informative for certain traditional NLP tasks, but they can still be essential for meaning and context.**

## 7. One more important point

Stop-word removal is different from tokenization.

Tokenization:

```
"The cat is sleeping"
        ↓
["The", "cat", "is", "sleeping"]
```

Stop-word removal:

["The", "cat", "is", "sleeping"]
        ↓
["cat", "sleeping"]

So:

**Tokenization → Break text into tokens**

**Stop-word removal → Remove selected tokens**

## ✅Key Takeaways

- **Stop words** are very common words that may provide little value for some NLP tasks.
- Traditional NLP often removed them to reduce unnecessary information.
- Stop-word removal can be useful with approaches like **Bag of Words and TF-IDF.**
- Don't blindly remove words such as "**not**", "**no**", "**never**" because they can change meaning.
- Modern LLM pipelines generally **do not perform traditional stop-word removal.**
- Whether a word is a "stop word" is **task-dependent.**

# NLP — 05. Stemming & Lemmatization

When working with text, the same word can appear in many different forms.

For example:
```
play
plays
played
playing
```

A traditional NLP system may want to recognize that these words are related to the same basic concept.

This is where stemming and lemmatization come in.

## 1. Why do we need them?

Consider these sentences:

`"The boy is playing football."`

`"The boys played football."`

`"The boy plays football."`

A simple system may see:

`playing`

`played`

`plays`

as three different words.

But semantically, they are all forms of:

**`play`**



There are two major approaches:

- **Stemming**
- **Lemmatization**

## 2. Stemming

**Stemming removes prefixes or suffixes from a word to obtain a base-like form.**

The important thing is:

> A stemmer does not necessarily produce a valid English word.

It generally works using simple rules rather than understanding the actual meaning of the word.

Example

```
playing → play
played  → play
plays   → play
```

But it can also produce something like:

`studies → studi`

`"studi"` isn't a proper English word.

That's okay for a stemmer. Its goal is mainly to reduce related words to a common form.

````python
import nltk
from nltk.stem import PorterStemmer #Porter Stemmer is a popular stemmer from NLTK.

stemmer = PorterStemmer()

words = ["playing", "play", "run", "running", "programming", "beginning", "happiness", "studying"]

for word in words:
    print(word, "→", stemmer.stem(word))
````

The stemmer has grouped them together, but **"happi" & "studi" is not a real English word.**

That's one of the main limitations of stemming.

## 3. Lemmatization

**Lemmatization is the process of converting a word into its correct base or dictionary form, called its lemma.**

For example:

```
playing  → play
played   → play
studies  → study
cars     → car
better   → good
```

The important thing is:

> **Lemmatization tries to produce a real, meaningful word.**

This is different from stemming, which simply tries to chop a word down.

#### How is it different from Stemming

Suppose we have: `studies`

**Stemming**

A stemmer may simply remove characters: `studies` → `studi`

`studi` isn't an English word.

**Lemmatization**

Lemmatization understands that: `studies` → `study`

`study` is the correct base word.

```
Stemming
→ "Give me a shortened form."

Lemmatization
→ "Give me the correct dictionary form."
```

````python
import nltk
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "studies", "cars", "runner", "running", "computer"]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word))
    
#You might be surprised by the output.
````

Why didn't `"playing"` become `"play"`?

Because **lemmatization needs grammatical information** to determine how a word is being used.

## 4. Stemming vs Lemmatization

Consider: `"studies"`

**Stemming** - `studies → studi`

**Lemmatization** - `studies → study`

So:

```
Stemming
→ Fast
→ Simple
→ Rule-based
→ May produce invalid words

Lemmatization
→ More linguistically accurate
→ Produces valid base words
→ Usually more computationally expensive
```

Remember this

> **Stemming: chopping a word down**

> **Lemmatization: finding the correct base word**

## 5. Part of Speech (POS) matters

Consider: `"playing"`

It could be a verb:

> "I am playing football."

or potentially part of a noun phrase:

> "Playing football is fun."

The lemmatizer can produce better results when we tell it the **part of speech.**

For example:

````python
lemmatizer.lemmatize("playing", pos="v")
````

````python
lemmatizer.lemmatize("better", pos="a")
````

````python
#v → verb
#a → adjective
#n → noun
#r → adverb
````

This shows why lemmatization is more sophisticated than simply removing characters.

## 6. Do modern LLMs use stemming or lemmatization?

Generally, **no—not as a preprocessing step.**

This is important because we're learning these concepts primarily to understand the evolution of NLP.

Traditional NLP pipeline:

```
Raw Text
   ↓
Tokenization
   ↓
Stop-word Removal
   ↓
Stemming / Lemmatization
   ↓
BoW / TF-IDF
   ↓
ML Model
```

Modern LLM pipeline:

```
Raw Text
   ↓
Tokenizer
   ↓
Token IDs
   ↓
Embeddings
   ↓
Transformer
```

LLMs need the original word forms because **word form and context can carry information.**

For example:

```
"The company is growing."
"The company grew rapidly."
"The company will grow."
```

The differences between `growing`, `grew`, and `grow` can matter to the model.

So we generally don't reduce everything to:

```
grow
grow
grow
```

before passing it to an LLM.

## 7. When are stemming and lemmatization useful?

They are particularly useful when you're building **traditional NLP systems** where reducing vocabulary size is beneficial.

Examples:

- Search systems
- Text classification
- Document clustering
- Information retrieval
- Keyword analysis

For example, if a search system receives:

> "I am studying machine learning."

and a document contains:

> "Students study machine learning."

Normalizing:

`studying → study`
`study → study`

can help the system recognize the relationship.

Modern search systems can use much more sophisticated semantic techniques, but understanding this traditional approach is important
